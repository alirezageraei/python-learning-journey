## Regex
String and texts master method is here!

```python
### regular expressions : Regex ###
"""
Regex is a search pattern for strings
Its regular application should be one of these:
1.finding specific part of a passage
2.collect them
3.replace or use them
"""
import re

passage = 'Hi, I am ali and this is a test for regex search pattern\nmy nubmer is +989182525255 and my nationality number is 3950645895'

r = re.search('ali',passage) # shows me obj, start and end index, match
re.search('sali',passage) # what about one character more?
re.search('li',passage) # or one less?
r.start() # you saw that? '''the method .search() would only show you one and first match pattern'''
r.group() # works with () groups
r.span() # shows the start and end index

findall = re.findall('ali', passage) # finds all match and put them in a list
find_iter = re.finditer('ali', passage) # insted of list, it returns a iterator that can be used in loops
for i in find_iter:
    ''' exp for finditer '''
    print(i)
```

### How to use Regex properly?
http://regex101.com --> best way to understand Regex

```python
import re

alireza = 'Alireza, +9891865489, 1021689547'
display(re.findall(r'\+\d+', alireza))
# \+ for + || \d for digit || + for more than 1 -->(result) get more than 1 digit after +
display(re.findall(r'\+\d{2, 4}', alireza))
# only digits with len between 2, 4 (after plus sign)
display(re.findall(r'\+\d*', alireza))
# + gets 0 or more after see + || * gets 0 or more after see +
display(re.findall(r'\+\d?', alireza))
# minimum 0 and maximum 1
display(re.findall(r'^\w+', alireza))
display(re.findall(r'\d{5}$', alireza))
# ^ gets only starting characters
# $ gets only ending characters (for example: getting files extentions)
```

```python
### cleaner Regex ###
import re

data = """
Alireza, +98918626563, @Arggera, 1236-6584-3248-5565
Fateme, +98903659865, @FUT5486, 2548-3458-9966-3548
Sadra, +98901359652, @sadra_64, 3458-5648-6648-2157
"""
# we can make a pattern (by method .compile()) then use it
pattern = re.compile(r'\d{4}-\d{4}-\d{4}-\d{4}')
re.findall(pattern, data)

# we also can make specific outputs of patterned datas
new_pattern = re.compile(r'(\d{4})-\d{4}-\d{4}-\d{4}') # returns only the first column
also_new_pattern = re.compile(r'(\d{4})-(\d{4})-\d{4}-\d{4}') # returns 2 first column 
re.findall(new_pattern, data)

# and we can do grouping
# grouping is a very useful method that can be used in
# seperating data's and show just that part we want it to show 
new_pattern = re.compile(r'(\d{4})-(\d{4})-(\d{4})-(\d{4})') # returns only the first column
data_iter = re.finditer(new_pattern, data)
for i in data_iter:
    print(i.group())
    print(i.group(1))
    print(i.group(2)) # unlike lists, index are stars from 1 not 0
```

```python
### practice group ###
### show all IDs without their starting @ sign ###
import re

data = """
Alireza, +98918600026563, @Arggera, 1236-6584-3248-5565
Fateme, +98903650009865, @FUT5486, 2548-3458-9966-3548
Sadra, +98901359000652, @sadra_64, 3458-5648-6648-2157
"""
pattern_1 = re.compile(r'\@\w+') # shows @
pattern_2 = re.compile(r'\@(\w+)') # groups the part after @

find_1 = re.finditer(pattern_1, data)
find_2 = re.finditer(pattern_2, data)

# shows @
for item in find_1:
    print(item.group())

# show ID without @
for item in find_2:
    print(item.group(1))
```
