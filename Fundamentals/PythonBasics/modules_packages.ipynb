```python
from datetime import datetime as dt

def week_day(date_string):
    # تبدیل رشته به شیء datetime
    date_obj = dt.strptime(date_string, "%Y-%m-%d")
    # برگرداندن نام روز هفته
    return date_obj.strftime("%A")  # "%A" برای نام کامل روز (مثلاً Monday)

# تست
d = week_day("2026-07-20")
print(d)  # Monday
```
