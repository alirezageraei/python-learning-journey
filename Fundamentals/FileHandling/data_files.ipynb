# csv

```python
import csv

my_file = open('Data/mycsv.csv', encoding="utf-8")
my_file.read()
my_file.close()
```

```python
my_file = open('Data/mycsv.csv', encoding="utf-8")
csv_reader = csv.reader(my_file)
# for i in csv_reader:
#     print(i)
my_file.close()
```

```python
with open('Data/mycsv.csv', encoding="utf-8") as my_file:
    data = csv.reader(my_file)
    next(data, None) # skips first row (Headers)
    lines = list(data)


s = {}
for line in lines:
    model = line[1]
    s[model] = s.get(model, 0) +1
# this line was: s
my_file.close()
```

also we can read our files as dictionary like this

```python
my_file = open('Data/mycsv.csv', encoding="utf-8")
data_dict = csv.DictReader(my_file)
lines = list(data_dict)
# this line was: lines
```

CSV Writer

```python
output = open('Data/mycsv.csv', mode='w', newline='')
my_writings = csv.writer(output, delimiter='|')
my_writings.writerow(['alireza','hossein', 'morteza'])
output.close()
```

# Excel

```python
import openpyxl

wb = openpyxl.load_workbook('Data/myExcel.xlsx')  # workbook: entire Excel file
ws = wb.active  # workSheet: a sheet in workbook (فعال‌ترین شیت)

ws['B6']
ws['C1']        # we can call cells
this_cell = ws['A4']  # point at A4 cell
this_cell.fill
this_cell.value
```

```python
t = 0
for i in range(2,10):
    cell_address = f"H{i}"
    this_cell = ws[cell_address]
    t += this_cell.value
t
```

making new Excel file

```python
new_wb = openpyxl.Workbook()
ws = new_wb.active
ws["A1"] = "Alireza"
ws["B1"] = 23
ws["A2"] = "Sara"
ws["B2"] = 20
ws.append(["John", 25])
new_wb.save
```
