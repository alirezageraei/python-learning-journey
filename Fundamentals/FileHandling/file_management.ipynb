# open, read and with

```python
l = ['ali','hasan','saman','zahra','sara']

with open("names.txt", "w") as n:
    for i in l:
        n.write(f"{i}\n")
```

```python
with open("names.txt", "r")as n:
    # f.read()
    for line in n:
        print(line, end="")
```

```python
with open("names.txt", "a") as n:
    n.write("Hello, Worlds")
```

# Json

```python
import json

books= [
        {"name": "History",
        "author": "Herodote",
        "year": 650},
        {"name": "Quantum",
        "author": "Halidy",
        "year": 1995},
        {"name": "Physics",
        "author": "Halidy",
        "year": 1970},
        {"name": "Cpp Programming",
        "author": "Robert j.roskey",
        "year": 2001}
]
```

```python
with open("Books.json", "w") as b:
    json.dump(books, b, indent=4)
```

```python
with open("Books.json", "r") as b:
    y = json.load(b)
    for book in y:
        print(book, end="\n"+"*"*50+"\n")
```

```python
auth = input("Author name: ").lower()
with open("Books.json", "r") as b:
    data = json.load(b)
    found = False
    for i in data:
        if i["author"].lower() == auth:
            print(i["name"])
            found = True
    if not found:
        print("Not found")
```

# CSV

```python
import csv

grades = [[20,19,17,19,20],[13,20,20,20,16],[10,9,12,16,19]]

with open("grades.csv", "w", newline="") as g:
    writer = csv.writer(g)
    writer.writerows(grades)
```

```python

with open("grades.csv", "r", newline="") as g:
    data = csv.reader(g)
    all_numbers = []
    for lst in data:
        if lst:  # اگر لیست خالی نبود
            for i in lst:
                all_numbers.append(int(i))
    
    total = sum(all_numbers)
    count = len(all_numbers)
    average = total / count if count > 0 else 0
```

```python
print(f"Numbers: {all_numbers}")
print(f"Total sum: {total}")
print(f"Count: {count}")
print(f"Average: {average:.2f}")
```
