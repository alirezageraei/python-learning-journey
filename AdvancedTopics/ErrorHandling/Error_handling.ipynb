### 2. try-except in Function
Handling ZeroDivisionError.

```python
### Try - Except in function ###

def divide(a, b):
    try:
        return a / b
    except ZeroDivisionError as e:
        print(f"Error occurred: {e}\n")
        return None

num = 3.141592653592
sub = float(input("Enter sub division: "))
res = divide(num, sub)
print(res)
```

### 4. Multiple Exceptions
Catching IndexError and TypeError.

```python
def value_by_index(lst, index):
    try:
        return lst[index]
    except IndexError:
        print("Index Out of Range")
    except TypeError:
        print("Input must be a list")

l = ['Alireza','Hossein','Negar','Sara','Kosar']
i = int(input(f"Enter the index: "))

value_by_index(l,i)
```

### 5. else and finally
Counting non-empty lines with proper file closing.

```python
### else and finally ###

def read_files(filename):
    f = None
    not_empty_lines = 0
    try:
        f = open(filename, "r")
    except FileNotFoundError:
        print("file not found!")
        return None
    else:
        for line in f:
            if line.strip():
                not_empty_lines += 1
    finally:
            if f is not None:
                f.close()
    return not_empty_lines
read_files("grades.csv")
```

### 6. raise Statement
Password length validator.

```python
### raise ###
def pass_length(password):
    length = len(password)
    if length < 8:
        raise ValueError("password must be at least 8 character")
    return True
    
passkey = input("Type in your password: ")

try:
    if pass_length(passkey):
        print(f"password is valid ✅'{passkey}'")
except ValueError as e:
    print(f"{e}")
```

### 7. Stack Trace Propagation
Catching, logging, and re-raising exceptions.

```python
### Stack trace ###

"""
    A method to show all Errors
    in potential Error spots
    like function input values
"""

from math import sqrt

def root(number):
    try:
        return sqrt(number)
    except ValueError as e:
        print(f"Caught error: {e}")  # ثبت خطا (اختیاری)
        raise  # ✅ دوباره همان خطا را صادر می‌کند

# بخش اصلی
n = int(input("Enter your number: "))

try:
    result = root(n)
    print(f"Result: {result}")
except ValueError as e:
    print(f"Final error caught: {e}")
```

### 8. Custom Exception
BankAccountError for withdrawal validation.

```python
### Custom Exception ###
# Custom exceptions are usually empty (just 'pass')
# But we can add extra logic if needed

class BankAccountError(Exception):
    pass

class BankAccount:
    def __init__(self, balance=0):
        self.balance = balance
    
    def withdraw(self, amount):
        if amount > self.balance:
            raise BankAccountError(f"your balance is not enough: {self.balance}")
        self.balance -= amount
        return self.balance

account = BankAccount(1000)
try:
    account.withdraw(2000)
except BankAccountError as e:
    print(e)
```

### 9. try-finally Pattern
Manual resource management simulation.

```python
### try-finally Pattern ###

class ManagedFile:

    """
        This is an example for understanding 'finally' better.
    """

    def __init__(self, filename):
        self.filename = filename
        self.file = None

    def open(self):
        self.file = open(self.filename, "r")
        return self.file
    
    def close(self):
        if self.file:
            self.file.close()

grades = "grades.csv"
grades_ = ManagedFile(grades)
try:
    grades_.open()
except  FileNotFoundError as e:
    print(e)
finally:
    grades_.close()
```
