## Why did I use matrixay instead of List?


Basis|List|matrixay
| --- | --- | --- |
Data Type Handling|Can store mixed data types|Stores only one data type
Memory Usage|Uses more memory due to flexible storage|Uses less memory because of uniform data type
Performance|Slower for numerical operations|Faster for numeric computations
Flexibility	Supports|easy insertion, deletion and resizing|Less flexible due to fixed type and size constraints
Arithmetic Operations|Cannot perform element-wise arithmetic directly|Supports element-wise arithmetic
Built-in Support|Available by default in Python	Requires|importing the matrixay module
Best Use Case|General-purpose and mixed-type data|Numeric and scientific data processing
Data Consistency|Allows mixed values without restriction|Enforces strict type consistency

```python
from itertools import combinations, product

def submatrix(matrix):
    if not isinstance(matrix, np.ndarray):
        matrix = np.array(matrix)
    
    if not np.all((matrix == -1) | (matrix == 0) | (matrix == 1)):
        return "This matrix is not Totally Unimodular because It has at least 1 value thats not in [-1, 0, 1]."
    
    k = min(matrix.shape)
    count = 0
    for i in range(1, k+1):
        rows_comb = combinations(range(matrix.shape[0]), i)
        cols_comb = combinations(range(matrix.shape[1]), i)
        
        for rows, cols in product(rows_comb, cols_comb):
            count += 1
            yield matrix[np.ix_(rows, cols)]
```

## محاسبه دترمینان ها

```python
def determinant_laplace(matrix):
    
    n = matrix.shape[0]
    if n == 1:
        return matrix[0, 0]
    if n == 2:
        return matrix[0,0]*matrix[1,1] - matrix[0,1]*matrix[1,0]
    det = 0
    for j in range(n):
        minor = np.delete(np.delete(matrix, 0, axis=0), j, axis=1)
        sign = -1 if j % 2 else 1
        det += sign * matrix[0, j] * determinant_laplace(minor)
    return det
    if det not in [-1, 0, 1]:
        return "This matrix is not Totally Unimodular because It has at least 1 submatrix determinant thats not in [-1, 0, 1]."
    return det
```
