# *Numpy*

#### Universal Functions ```Ufunc```

````python
import numpy as np
````

````python
x = np.arange(27).reshape(3,3,3)
np.add(x, 1)
np.add.reduce(x, 1) 
np.shape(np.add.reduce(x, 1))

np.add.reduce(x, (1, 2))
# np.shape(np.add.reduce(x, (1, 2)))
````

````python
x.dtype
np.multiply.reduce(x, axis=0,dtype=float)
````

````python
x = np.arange(9).reshape(3,3)
y = np.zeros(3, dtype=int)

np.multiply.reduce(x, axis=1, dtype=float, out=y)
# keyword 'out' will set all x multiplied values to y | consider axis
y
````

````python
mark = {False : "  -", True : "  Y"}
def print_table(ntypes):

    print("X  " + '  '.join(ntypes))
    for row in ntypes:
        print(row, end='')
        for col in ntypes:
            print(mark[np.can_cast(row, col)], end='')
        print()

print_table(np.typecodes['All'])
````
