### Replicating, joining, or mutating existing arrays

```python
# Replicate by copy
arr1 = np.copy(array)
arr2 = arr1[1:2].copy()
arr2 *= 5

display(arr1)
display(arr2)
```

```python
# join by block
A = np.ones((2, 2))
B = np.eye(2, 2)
C = np.zeros((2, 2))
D = np.diag((-3, -4))
# to see how its works
display(np.block([[A, B], [C, D]]))
display(np.block([[C, B, C], [C, D, C]]))

""" Also read abut hstack and vstack """
```
