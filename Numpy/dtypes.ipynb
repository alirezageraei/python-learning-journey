a point to prevent overflow: ```casting=""```

````python
a = np.array([300] , dtype=np.int64)
a.astype(np.int8) # returns 44 in result, reason? overflow (int8 supports numbers to 255-6, so 300 - 256)
a.astype(np.int8, casting="same_kind")
````

````python
casting_values = [
    "no",         # هیچ تبدیلی مجاز نیست
    "equiv",      # فقط تبدیل‌های معادل (مثل تغییر byte order)
    "safe",       # فقط تبدیل‌هایی که مقدار را حفظ می‌کنند
    "same_kind",  # تبدیل بین نوع‌های هم‌خانواده
    "unsafe"      # هر تبدیلی مجاز است (پیش‌فرض)
]

arr = np.array([300], dtype=np.int64)
try:
    arr.astype(np.int8, casting='no')        # Error
    arr.astype(np.int8, casting='equiv')     # Error
    arr.astype(np.int8, casting='safe')      # Error
    arr.astype(np.int8, casting='same_kind') # succsses but value changes (overflow)
    arr.astype(np.int8, casting='unsafe')    # succsses but value changes (overflow)
except TypeError as e:
    print(e)
````

#### dtypes(summarized):

````python
# ==================================================
# آموزش کامل دیتاتایپ‌های NumPy (int8, uint32, ...)
# ==================================

print("="*60)
print("تفاوت int8 و uint8 (محدوده مقادیر)")
print("="*60)

# int8: -128 تا 127
int8_arr = np.array([-128, 0, 127], dtype=np.int8)
print(f"int8 array: {int8_arr}")
print(f"محدوده int8: از -128 تا 127\n")

# uint8: 0 تا 255
uint8_arr = np.array([0, 128, 255], dtype=np.uint8)
print(f"uint8 array: {uint8_arr}")
print(f"محدوده uint8: از 0 تا 255\n")

# ==================================================
print("="*60)
print("سرریز (Overflow) - وقتی از محدوده خارج بشیم")
print("="*60)

# مثال سرریز در int8
x = np.array([30], dtype=np.int8)
print(f"قبل از تغییر: {x[0]}")
x[0] = 150  # 150 از محدوده int8 خارج است
print(f"بعد از اختصاص 150 به int8: {x[0]} (سرریز شده!)")
print("چرا؟ چون 150 - 256 = -36\n")

# مثال سرریز در uint8
y = np.array([200], dtype=np.uint8)
print(f"قبل از تغییر: {y[0]}")
y[0] = 300  # 300 از محدوده uint8 خارج است
print(f"بعد از اختصاص 300 به uint8: {y[0]} (سرریز شده!)")
print("چرا؟ چون 300 - 256 = 44\n")

# ==================================================
print("="*60)
print("مقایسه حافظه اشغالی (int8 vs int32 vs int64)")
print("="*60)

arr_int8 = np.array([1, 2, 3, 4, 5], dtype=np.int8)
arr_int32 = np.array([1, 2, 3, 4, 5], dtype=np.int32)
arr_int64 = np.array([1, 2, 3, 4, 5], dtype=np.int64)

print(f"int8  - حافظه: {arr_int8.nbytes} بایت  (هر عدد 1 بایت)")
print(f"int32 - حافظه: {arr_int32.nbytes} بایت (هر عدد 4 بایت)")
print(f"int64 - حافظه: {arr_int64.nbytes} بایت (هر عدد 8 بایت)")
print(f"آرایه یکسان با 5 عضو، ولی حافظه {arr_int64.nbytes / arr_int8.nbytes} برابر بیشتر!")

# ==================================================
print("\n" + "="*60)
print(" استفاده درست - کِی از کدوم dtype استفاده کنیم؟")
print("="*60)

# مثال: پردازش تصویر (پیکسل‌ها بین 0 تا 255)
image_pixels = np.array([[100, 150, 200], 
                          [50, 75, 255]], dtype=np.uint8)
print(f"پیکسل‌های تصویر:\n{image_pixels}")
print(f"نوع داده مناسب برای تصویر: {image_pixels.dtype}\n")

# مثال: سن افراد (هیچ‌وقت منفی نیست)
ages = np.array([25, 32, 18, 45, 67], dtype=np.uint8)
print(f"سن افراد: {ages}")
print(f"نوع داده: {ages.dtype} (کافی و بهینه)\n")

# مثال: دمای زیر صفر (نیاز به منفی داریم)
temperatures = np.array([-5, 0, 12, -3, 8], dtype=np.int8)
print(f"دما (با منفی): {temperatures}")
print(f"نوع داده: {temperatures.dtype}\n")

# ==================================================
print("="*60)
print("تبدیل دیتاتایپ (casting)")
print("="*60)

original = np.array([10, 20, 30], dtype=np.int16)
print(f"اصلی: {original} - dtype: {original.dtype}")

# تبدیل به uint8
converted = original.astype(np.uint8)
print(f"تبدیل به uint8: {converted} - dtype: {converted.dtype}")

# تبدیل به float32
float_version = original.astype(np.float32)
print(f"تبدیل به float32: {float_version} - dtype: {float_version.dtype}\n")

# ==================================================
print("="*60)
print("جدول خلاصه دیتاتایپ‌های پرکاربرد")
print("="*60)

print("""
┌─────────┬─────────────────────┬─────────────────────────┐
│ dtype   │ معنی                │ محدوده تقریبی           │
├─────────┼─────────────────────┼─────────────────────────┤
│ int8    │ صحیح علامت‌دار ۸ بیت │ -128 تا 127             │
│ int16   │ صحیح علامت‌دار ۱۶ بیت│ -32,768 تا 32,767       │
│ int32   │ صحیح علامت‌دار ۳۲ بیت│ -2.1 میلیارد تا 2.1 میلیارد│
│ int64   │ صحیح علامت‌دار ۶۴ بیت│ خیلی بزرگ (9.2×10¹⁸)    │
│ uint8   │ صحیح بدون علامت ۸ بیت│ 0 تا 255                │
│ uint16  │ صحیح بدون علامت ۱۶ بیت│ 0 تا 65,535            │
│ uint32  │ صحیح بدون علامت ۳۲ بیت│ 0 تا 4.29 میلیارد      │
│ float32 │ اعشاری ۳۲ بیت        │ دقت حدود ۷ رقم اعشار    │
│ float64 │ اعشاری ۶۴ بیت        │ دقت حدود ۱۵ رقم اعشار   │
└─────────┴─────────────────────┴─────────────────────────────┘
┘
""")

# ==================================================
print("="*60)
print("نکات طلایی برای انتخاب dtype:")
print("="*60)
print("""
1. اگر مقدار منفی دارید → از int استفاده کنید (نه uint)
2. اگر فقط اعداد مثبت و صفر دارید → uint حافظه بهتری می‌دهد
3. برای تصاویر و ویدیو (0-255) → uint8 عالی است
4. برای محاسبات علمی دقیق → float64 پیش‌فرض خوبی است
5. همیشه کوچکترین d-type ممکن را انتخاب کنید تا حافظه ذخیره شود
6. مواظب سرریز باشید! NumPy خطا نمی‌دهد، بی‌صدا عدد را دور می‌زند
""")
````
