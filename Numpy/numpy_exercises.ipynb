## 3. View vs Copy
Slicing returns views (modifies original). Integer indexing returns copies. Use `.copy()` for explicit copy.

```python
### view and copy ###
arr = np.arange(25, dtype=np.int8).reshape(5, 5)
third_row = arr[2, :]
second_col = arr[:, 1]

print("this is a view" if third_row.base is not None else "this is a copy")
print("this is a view" if second_col.base is not None else "this is a copy")
print('-'*20)
# if we want a copy we can do this
third_row = arr[2, [0, 1, 2, 3, 4]]
second_col = arr[[0, 1, 2, 3, 4], 1]

#or even more s1mple
third_row = arr[2, :].copy()
second_col = arr[:, 1].copy()

print("this is a view" if third_row.base is not None else "this is a copy")
print("this is a view" if second_col.base is not None else "this is a copy")
```

## 7. Extract rectangle

## 10. Student degree matrix
Calculate per-student mean/max/min from 30×5 matrix. Display pass/fail (mean ≥ 10).

```python
### students degree matrix ###

students_degree = rng.integers(0, 21, 150).reshape(30,5)

student_mean_degree = np.mean(students_degree, axis=1)
max_degree = np.max(students_degree, axis=1)
min_degree = np.min(students_degree, axis=1)

counter = 0
for student in students_degree:
    print(
        f"ID : {counter + 1} | "+
        f"degrees mean : {student_mean_degree[counter]} | "+
        f"maximum degree : {max_degree[counter]} | "+
        f"minimum degree : {min_degree[counter]} | ", end='')
    print("Passed!" if student_mean_degree[counter] >= 10 else "Failed")
    
    counter += 1
```

## 11. Filter high-performing students
Find students with mean degree ≥ 15 using loop over array.

```python
counter = 0
for student in students_degree:
    if student_mean_degree[counter] >= 15:
        print(f"student with ID {counter + 1}, has mean degrees higher than 15")
    counter += 1
```

## 17. Moving average
Using cumulative sum trick: `(cumsum[window:] - cumsum[:-window]) / window`.

```python
### moving average ###

def moving_average(data, window=5, axis=0):

    res = np.cumsum(data, axis=axis, dtype=np.float32)
    moving = (res[window:] - res[:-window]) / window
    return moving

arr = np.arange(1000)
moving_average(arr)
```

## 18. Find anomalies (method 1)
Rows where `(mean - std)` is below global average of this metric.

```python
### Find Anomaly ###

arr = rng.integers(0, 100, 2500).reshape(50,50)

mean = np.mean(arr, axis=1)
std = np.std(arr, axis=1)

expected_mean = np.mean(mean)
expected_std = np.mean(std)

for row in arr:
    if expected_mean - expected_std > np.mean(row) - np.std(row):
        print(row)
```

## 19. Find anomalies (method 2 - Z-score)
Standardize row means using Z-score, find rows with |Z| > 2 (statistical outliers).

```python
arr = rng.integers(0, 100, 2500).reshape(50,50)

mean = np.mean(arr, axis=1)

z = (mean - np.mean(mean)) / np.std(mean)
z[np.abs(z) > 2]
```

## 20. Distance between all points (no loops)
Pairwise distance matrix for 100 2D points. Broadcasting: reshape to (100,2,1) and (1,2,100) → diff → sum → sqrt.(Broadcasting function)

```python
### distance between all points (without loops) ###

points = rng.integers(0, 10, 200, dtype=np.int8).reshape(100, 2)

arr1 = points[:, :, np.newaxis]
arr2 = points.T[np.newaxis, :]

res = np.sqrt(np.sum((arr1 - arr2)**2, axis=1))
res
```
