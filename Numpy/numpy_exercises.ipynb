## 1. Making an array
Create array 1-100, filter even numbers using boolean indexing.

```python
### Making an array ###
arr = np.arange(1,101, dtype=np.int8)
evens = arr[arr %2 == 0] # this is how to use boolean as filter
evens
```

## 2. Random array
Generate random ints (1-100). Set values below mean to zero using boolean mask.

```python
### Random array ###
rng = np.random.default_rng(42) # seed makes same random cycle in other devices

arr = rng.integers(1, 101, 100, dtype=np.int8)
arr[arr < arr.mean()] = 0
arr
```

## 3. View vs Copy
Slicing returns views (modifies original). Integer indexing returns copies. Use `.copy()` for explicit copy.

```python
### view and copy ###
arr = np.arange(25, dtype=np.int8).reshape(5, 5)
third_row = arr[2, :]
second_col = arr[:, 1]

print("this is a view" if third_row.base is not None else "this is a copy")
print("this is a view" if second_col.base is not None else "this is a copy")
print('-'*20)
# if we want a copy we can do this
third_row = arr[2, [0, 1, 2, 3, 4]]
second_col = arr[[0, 1, 2, 3, 4], 1]

#or even more s1mple
third_row = arr[2, :].copy()
second_col = arr[:, 1].copy()

print("this is a view" if third_row.base is not None else "this is a copy")
print("this is a view" if second_col.base is not None else "this is a copy")
```

## 4. Reverse
Reverse arrays using `np.flip()` or slicing `[::-1]` on any dimension.

```python
### Reverse ###
arr = np.arange(20, dtype=np.int8).reshape(2,2,5)
np.flip(arr) # reversed by flip func
arr[::-1] # reversed by slicing
```

## 6. Simple broadcasting
Add shapes (5,) and (7,1) → result (7,5). NumPy automatically expands dimensions.

also, we can do like this: (normal standard)

```python
### s1mple broadcasting ###
arr1 = np.arange(5)
arr2 = np.arange(7).reshape(7,1)

res = arr1 + arr2 # answer will be (7, 5)
res
```

## 7. Extract rectangle
Select non-contiguous rows & columns. Two methods: chained indexing or `np.ix_()`.

```python
### extract a rectangle without np.ix_ ###
arr = np.arange(36).reshape(6, 6)

rows = [1, 3, 5]
columns = [0, 2, 4]

res = arr[rows, :][:, columns].copy()
print(res, "\n")

# also: with np.ix_
res = arr[np.ix_(rows, columns)]
print(res)
```

## 8. Operate based on axis
Sum rows (axis=1), filter rows where sum > 100. Shows axis-aware operations.

```python
### s1mple operate based on axis ###
arr = rng.integers(0,40,25).reshape(5,5)
rows = np.sum(arr, 1) > 100
print(arr,'\n'+'-'*20)
print(arr[rows])
```

## 9. Boolean masking
Replace evens with 0, odds with 1. Alternative: `np.where()` for cleaner code.

```python
### boolean masking ###

arr = rng.integers(0, 1000, 10000, dtype=np.uint64).reshape(100, 100)
evens = arr %2 == 0
odds = arr %2 == 1
arr[evens] = 0
arr[odds] = 1
# better way:
# arr[:] = np.where(arr % 2 == 0, 0, 1)
print(arr)
```

## 10. Student degree matrix
Calculate per-student mean/max/min from 30×5 matrix. Display pass/fail (mean ≥ 10).

```python
### students degree matrix ###

students_degree = rng.integers(0, 21, 150).reshape(30,5)

student_mean_degree = np.mean(students_degree, axis=1)
max_degree = np.max(students_degree, axis=1)
min_degree = np.min(students_degree, axis=1)

counter = 0
for student in students_degree:
    print(
        f"ID : {counter + 1} | "+
        f"degrees mean : {student_mean_degree[counter]} | "+
        f"maximum degree : {max_degree[counter]} | "+
        f"minimum degree : {min_degree[counter]} | ", end='')
    print("Passed!" if student_mean_degree[counter] >= 10 else "Failed")
    
    counter += 1
```
