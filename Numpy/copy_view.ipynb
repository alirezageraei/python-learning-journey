# *Numpy*
#### **Copy and View**

````python
import numpy as np
````

````python
x = np.arange(10)
y = x[1:3]  # creates a view
x[1:3] = [10, 11]
display(x)
display(y) # y will change because its a view
````

````python
x = np.arange(9).reshape(3, 3)
y = x[[1, 2]]
x[[1]] = [[0, 0, 0]]
y # y wont change because it's a copy
````

the rule is simple: slicing -> view | fancy indexing, boolean masking -> copy

````python
x[:] # view
x[[1,2]] # copy
x[x<5] # copy

x[:, 1] # view
x[:,[1]] # copy
x[:,[1,2]] # copy
````

### Results:

```.base``` can seperate view and copy  
  
for view: .base "returns base is original"  
for copy: .base "returns base is None"  
  
reshape - slice - transpose:    View     base is Original  
advanced indexing - copy - astype:     copy     base is None
